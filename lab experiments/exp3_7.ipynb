# Greedy Best-First Search (Prolog)

The code and sample query below are saved in the notebook so they appear in its preview. The query searches for a path from `a` to `d`.

```prolog
% Graph edges
edge(a, b).
edge(a, c).
edge(b, d).
edge(c, d).

% Heuristic values (distance to goal 'd')
h(a, 5).
h(b, 4).
h(c, 2).
h(d, 0).

% Search rules
bfs(Goal, Goal, [Goal]).
bfs(Current, Goal, [Current | Path]) :-
    edge(Current, Next),
    h(Next, H1),
    \+ (edge(Current, Other), h(Other, H2), H2 < H1), % Ensure no neighbor has a smaller H
    bfs(Next, Goal, Path).

% Input query
?- bfs(a, d, Path).
```
