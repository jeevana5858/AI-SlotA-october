# Tower of Hanoi in Prolog

This notebook contains a Prolog program that solves the Tower of Hanoi problem.


```prolog
hanoi(N) :-
    move(N, left, middle, right).

move(0, _, _, _) :- !.
move(N, A, B, C) :-
    N > 0,
    M is N - 1,
    move(M, A, C, B),
    write('Move disk from '),
    write(A),
    write(' to '),
    write(C),
    nl,
    move(M, B, A, C).

?- hanoi(2).
```
