```prolog
% Diet Suggestion System Based on Disease

% Disease and diet suggestions

diet(diabetes, 'Eat whole grains, green vegetables and low-sugar fruits').
diet(blood_pressure, 'Eat low-salt food, fruits and vegetables').
diet(obesity, 'Eat salads, fruits and low-calorie foods').
diet(anemia, 'Eat spinach, dates, beans and iron-rich foods').
diet(heart_disease, 'Eat oats, vegetables and low-fat foods').
diet(acidity, 'Eat bananas, oats and non-spicy foods').

% Disease descriptions

disease(diabetes, 'High blood sugar').
disease(blood_pressure, 'High blood pressure').
disease(obesity, 'Excess body weight').
disease(anemia, 'Low hemoglobin').
disease(heart_disease, 'Heart-related condition').
disease(acidity, 'Excess stomach acid symptoms').

% Suggestion rule

suggest_diet(Disease) :-
    disease(Disease, Description),
    diet(Disease, Food),
    write('Disease: '),
    write(Description), nl,
    write('Suggested Diet: '),
    write(Food), nl.

% Main program

main :-
    write('DIET SUGGESTION SYSTEM'), nl,
    write('Available Diseases:'), nl,
    write('1. diabetes'), nl,
    write('2. blood_pressure'), nl,
    write('3. obesity'), nl,
    write('4. anemia'), nl,
    write('5. heart_disease'), nl,
    write('6. acidity'), nl,
    nl,
    write('Enter disease name: '),
    read(Disease),
    suggest_diet(Disease).

:- initialization(main).
```
