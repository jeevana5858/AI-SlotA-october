```prolog
monkey(at_door).
box(at_window).
banana(at_middle).

can_get_banana :-
    monkey(at_door),
    write('Monkey walks to the box'), nl,
    write('Monkey pushes the box'), nl,
    write('Monkey climbs the box'), nl,
    write('Monkey takes the banana'), nl,
    write('Monkey got the banana!').
:- initialization(can_get_banana).
```
